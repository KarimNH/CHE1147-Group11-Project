To run this notebook, be sure that the model, 'train.csv', and 'test.csv' are in this folder too.

```python
import pandas as pd
import numpy as np
import joblib
import shap
shap.initjs() # For jupyter notebook
import matplotlib.pyplot as plt
from XAI import explanation_shap, per_imprt

import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
df_train = pd.read_csv('train.csv')
df_test = pd.read_csv('test.csv')
target_feature = "outputs.pbe.bandgap"
X_train = df_train.drop(columns=[target_feature,'MOFname'])
y_train = df_train[target_feature]
X_test = df_test.drop(columns=[target_feature,'MOFname'])
y_test = df_test[target_feature]
model_xgb = joblib.load('xgb_model.pkl')
```

```python
explainer_shap, shap_values, shap_df = explanation_shap(model = model_xgb,
                                x_train = X_train,
                                x_test = X_test,
                                task_type = "global",
                                index = 10,
                                show_plot = True)
```

```python
result_permtnew, importance_df_permtnew  = per_imprt(model = model_xgb,
                            x_test = X_test,
                            y_test = y_test,
                            n_repeats = 10,
                            show_plot = True,
                            plot_top_n = 10)
```
