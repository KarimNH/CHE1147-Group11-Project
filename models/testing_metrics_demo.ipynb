To run this notebook, be sure that "test.csv" is in this folder too.

```python
import pandas as pd
import joblib
from testing_metrics import get_regression_metrics
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
df_test = pd.read_csv('test.csv')
target_feature = "outputs.pbe.bandgap"
X_test = df_test.drop(columns=[target_feature,'MOFname'])
y_test = df_test[target_feature]
```

```python
dummyregressor_mean = joblib.load('dummyregressor_mean.pkl')
dummyregressor_mean_result = get_regression_metrics(dummyregressor_mean,X_test, y_test)
print("Dummy Model (Mean) Testing Metrics:")
print(dummyregressor_mean_result)

dummyregressor_median = joblib.load('dummyregressor_median.pkl')
dummyregressor_median_result = get_regression_metrics(dummyregressor_median,X_test, y_test)
print("Dummy Model (Median) Testing Metrics:")
print(dummyregressor_median_result)
```

```python
model_linreg = joblib.load('linear_reg.pkl')
linreg_result = get_regression_metrics(model_linreg,X_test, y_test)
print("Linear Regression Testing Metrics:")
print(linreg_result)
```

```python
model_krr_default = joblib.load('krr_model_default.pkl')
krr_result_default = get_regression_metrics(model_krr_default, X_test, y_test)
print("Default KRR Testing Metrics:")
print(krr_result_default)

model_krr = joblib.load('krr_best_model.pkl')
krr_result = get_regression_metrics(model_krr, X_test, y_test)
print("Tuned KRR Testing Metrics:")
print(krr_result)
```

```python
model_rf_default = joblib.load('rf_model_default.pkl')
rf_result_default = get_regression_metrics(model_rf_default,X_test, y_test)
print("Default Random Forest Testing Metrics:")
print(rf_result_default)

model_rf = joblib.load('rf_best_model.pkl')
rf_result = get_regression_metrics(model_rf,X_test, y_test)
print("Tuned Random Forest Testing Metrics:")
print(rf_result)
```

```python
model_xgb_default = joblib.load('xgb_model_default.pkl')
xgb_result_default = get_regression_metrics(model_xgb_default,X_test, y_test)
print("Default XGBoost Testing Metrics:")
print(xgb_result_default)

model_xgb = joblib.load('xgb_best_model.pkl')
xgb_result = get_regression_metrics(model_xgb,X_test, y_test)
print("Tuned XGBoost Testing Metrics:")
print(xgb_result)
```
